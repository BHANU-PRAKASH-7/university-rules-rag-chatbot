# University Rules Assistant — RAG Chatbot

## Project Overview
A Retrieval-Augmented Generation (RAG) chatbot that answers questions using a university rules document.

## Technologies Used
- Python
- Google Colab
- Sentence Transformers (`all-MiniLM-L6-v2`)
- Scikit-learn cosine similarity
- FLAN-T5-small
- Gradio

## How It Works
1. The university rules document is split into chunks.
2. Sentence Transformers converts each chunk into a numerical embedding.
3. When a user asks a question, the retriever finds the most similar chunk using cosine similarity.
4. A relevance threshold determines whether to answer or reject the question.
5. The answer generator produces a response, and Gradio displays the answer and retrieved rule.

## Current Features
- Answers questions about university rules.
- Retrieves relevant document passages.
- Displays the retrieved rule and similarity score.
- Rejects questions when the similarity score is below the configured threshold.
- Provides a web-based chat interface.

## Limitations
- Answer accuracy depends on the quality of the source document and retrieval.
- The similarity threshold is a heuristic and requires further evaluation.
- The application currently runs through Google Colab, and its shared link depends on the runtime remaining active.
- Generated answers should be checked against the retrieved university rules.

## System Architecture

1. **Knowledge source:** University rules are provided as plain text.
2. **Chunking:** The document is split into smaller text chunks.
3. **Embedding:** SentenceTransformer (`all-MiniLM-L6-v2`) converts each chunk into a 384-dimensional vector.
4. **Retrieval:** Cosine similarity identifies the chunk most relevant to the user's question.
5. **Answer generation:** FLAN-T5 (`google/flan-t5-small`) generates an answer using the retrieved context.
6. **Relevance check:** A similarity threshold helps reject questions that appear unrelated to the source document.
7. **User interface:** Gradio displays the answer, retrieved rule, and similarity score.

**Current limitation:** The system retrieves one best-matching chunk. Similarity thresholds are heuristic and do not guarantee factual correctness.
